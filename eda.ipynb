# Quick EDA: parquet dataset comparison

This notebook summarizes the main parquet datasets in the bucket and compares their schema, row counts, dtypes, and sample records.

```python
import glob
import os
import pandas as pd

patterns = {
    'digital_sessions': 'bucket/digital_sessions_*.parquet',
    'viewer_sessions': 'bucket/viewer_sessions_*.parquet',
    'viewer_weights': 'bucket/viewer_weights_*.parquet',
    'target_group_mappings': 'bucket/target_group_mappings_*.parquet',
}

summary = []
for name, pattern in patterns.items():
    files = sorted(glob.glob(pattern))
    if not files:
        continue
    df = pd.read_parquet(files[0])
    summary.append({
        'dataset': name,
        'file_count': len(files),
        'rows': df.shape[0],
        'columns': df.shape[1],
        'column_names': list(df.columns),
        'dtypes': df.dtypes.astype(str).to_dict(),
        'nulls': df.isna().sum().to_dict(),
        'sample': df.head(3).to_dict(orient='records'),
    })

summary
```

```python
digital_sessions_file = r"bucket/digital_sessions_000000000000.parquet"
df_digital = pd.read_parquet(digital_sessions_file)

print("Digital Sessions Summary")
print("=" * 50)
print(f"Shape: {df_digital.shape}")
print(f"\nColumn Names:\n{df_digital.columns.tolist()}")
print(f"\nData Types:\n{df_digital.dtypes}")
print(f"\nNull Counts:\n{df_digital.isna().sum()}")
print(f"\nFirst Few Rows:")
df_digital.head()
```

```python
viewer_sessions_file = r"bucket/viewer_sessions_000000000000.parquet"
df_viewer = pd.read_parquet(viewer_sessions_file)

print("Viewer Sessions Summary")
print("=" * 50)
print(f"Shape: {df_viewer.shape}")
print(f"\nColumn Names:\n{df_viewer.columns.tolist()}")
print(f"\nData Types:\n{df_viewer.dtypes}")
print(f"\nNull Counts:\n{df_viewer.isna().sum()}")
print(f"\nFirst Few Rows:")
df_viewer.head()
```
